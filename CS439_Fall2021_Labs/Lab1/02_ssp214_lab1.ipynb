### Linearly Spaced Points
There are many tasks that you need to generate a set of points in a range that are linearly spaced. For this task
we will use numpy modules arnage and linspace.

Note that the tutorial uses `np.arange`. While this is fine in some cases, we generally prefer to use `np.linspace`. `np.linspace(a, b, N)` divides the interval `[a, b]` into N points, while `np.arange(a, b, s)` will step from `a` to `b` with a fixed step size `s`.

One thing to keep in mind is that `np.linspace` always includes both end points while `np.arange` will *not* include the second end point `b`. For this reason, when we are plotting ranges of values we tend to prefer `np.linspace`.

Write two functions that return the same result.

```python
# create an array of linearly spaced points from -5(inclusive) to 6 (not inclusive) spaced by 1
list1 = np.arange(-5,6,1)
print(list1)

# do the same with linspace
np.linspace(-5,6,11, endpoint = False)
```

## Question 2.3 3D Plotting
The code below plot function f(x,y) in 3D. The function is 
$$f(x,y) = x^2 + y^2$$
### Activity 1
Observe the values of (x,y) that minimizes the function f(x,y)

Based on the generated plot, the values of (x,y) that minimize the function (x,y)  = (0,0).
### Activity 2
Change the function to $$f(x,y) = xy $$ and re-plot. Estimate the values (x,y) that minimizes f(x,y)?

Based on the generated plot, there are no exact values of (x,y) that minimize the function (x,y), but in general,
positive x-values and negative y values move towards an indefinite minimum.

```python
from mpl_toolkits import mplot3d
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return (x**2 + y**2)

x = np.linspace(-6, 6, 30)
y = np.linspace(-6, 6, 30)

X, Y = np.meshgrid(x, y)
Z = f(X, Y)

fig = plt.figure()

ax = plt.axes(projection='3d')
ax.contour3D(X, Y, Z, 50, cmap='binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

## Task 3 - Reading Documentation
We should also familiarize ourselves with looking up documentation and learning how to read it. Below is a section of code that plots a basic wireframe. Replace each `# Your answer here` with a description of what the line above does, what the arguments being passed in are, and how the arguments are used in the function. For example your answer may look like,

`np.arange(2, 5, 0.2)`

`# This returns an array of numbers from 2 to 5 with an interval size of 0.2`

Hint: The `Shift` + `Tab` (to see what a function does) tip from earlier in the notebook may help here.

```python
from mpl_toolkits.mplot3d import axes3d

u = np.linspace(1.5*np.pi, -1.5*np.pi, 100)
print(u)
# This will generate linearly spaced function vectors - there will be exactly 100 linearly spaced points
# between 1.5 pi and -1.5pi
[x,y] = np.meshgrid(u, u)

# returns 2-D grid coordinates based on the coordinates contained in vectors in the first and second arguments. 
# in this case the arguments both u -> so the coordinates returned are based on the vector u
# x is a matrix where each row is a copy of u, and y is a matrix where each column is a copy of u. 
# The grid represented by the coordinates x and y has length(u) rows and length(u) columns.

squared = np.sqrt(x.flatten()**2 + y.flatten()**2)
z = np.cos(squared)
# 1. Here, we're flattening the x matrix to one dimension and then squaring the values in it.
# 2. Next, we're doing the the same as above with the y matrix.
# 3. Then we add the resulting x and y matrices.
# 4. The square root function is applied to the resulting sum matrix and is assigned to the variable 'squared'
# 5. The mathematical function np.cos calculates trigonometric cosine for all the array elements in 'squared' and
# the resulting array is assigned to the variable 'z'

z = z.reshape(x.shape)
# The array z is being given a new shape - the dimensions of x - without changing its data.

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
# The figure object is created and assigned to the variable 'fig'. It has 0 sets of axes at first.
# A 3-D axis object is created by calling the add_subplot method and specifying the value ‘3d’ to the projection parameter.
# The first parameter - 111 - is a way of locating your subplot in the overall space of the figure object. 
# The code 111 can be though of as 1 row, 1 column, 1st position. 

ax.plot_wireframe(x, y, z, rstride=10, cstride=10)
# Here x and y are 2D array of x and y points and z is a 2D array of heights. 
# The keyword arguments rstride= and cstride= determine the row step size and the column step size, which in this case = 10.
# These keyword arguments control how close together the "wires" in the wire frame plot are drawn.

ax.view_init(elev=50., azim=30)
# Set the elevation and azimuth of the axes, used to rotate the axes programatically.
# ‘elev’ stores the elevation angle in the z plane, in this case 50 degrees. 
# ‘azim’ stores the azimuth angle in the x,y plane, in this case 30 degrees.

plt.savefig("figure1.png")
```
