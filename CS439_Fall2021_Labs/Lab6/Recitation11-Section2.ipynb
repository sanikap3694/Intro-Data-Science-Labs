```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = sns.load_dataset("iris")
data.head()
```

```python
data['species'].value_counts()
```

```python
y = data['total']
x = data[data.columns.difference(['total'])]
```

```python
print(y.shape)
print(x.shape)
```

```python
y.head()
```

```python
x.head()
```

```python
x = pd.get_dummies(x)
x
```

```python
def abs_error(y, y_hat):
    return np.abs(y-y_hat)
```

```python
from scipy.optimize import minimize

min_val = minimize(lambda theta: abs_error(x.dot(theta), y).mean(), x0=np.zeros(len(x.columns)))
```

```python
min_val["x"]
```

```python
yl = data['abbrev'].astype('category')
xl = data[data.columns.difference(['abbrev'])]
```

```python
yl = data['species'].astype('category')
xl = data[data.columns.difference(['species'])]
```

```python
print(xl.shape)
print(yl.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(xl,yl,test_size = 0.80, random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
print(y_train.value_counts())
print(y_test.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression

logmodel = LogisticRegression()
logmodel.fit(x_train, y_train)
```

```python
predictions = logmodel.predict(x_test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_true = y_test, y_pred = predictions))
```
