```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
# load 'car_crashes' dataset from seaborn library
data = sns.load_dataset("car_crashes")
print("Number of Records:", len(data))
data.head()
```

```python
data.columns.difference(['total'])
```

```python
# Linear Regression
# split the date into y('total') and x(all others)
y = data['total']
x = data[data.columns.difference(['total'])]
y
```

```python
x.head()
```

```python
# get one-hot encoding of x
x = pd.get_dummies(x)
x.head()
```

```python
# Split the data set to training and test set using sklearn train_test_split
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lm = LinearRegression()
model = lm.fit(x_train, y_train)
print(f'alpha = {model.intercept_}')
print(f'betas = {model.coef_}')
# y_train_pre = model.predict(X_train)
y_test_pre = model.predict(x_test)
```

```python
print(list(y_test))
print(list(y_test_pre))
```

```python
print(y_test-y_test_pre)
```

```python
import sklearn.metrics as metrics
print('MAE: ', metrics.mean_absolute_error(y_test, y_test_pre))
print('MSE: ', metrics.mean_squared_error(y_test, y_test_pre))
print('RMSE: ', np.sqrt(metrics.mean_squared_error(y_test, y_test_pre)))
```
