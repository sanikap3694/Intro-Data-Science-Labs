```python
# transform the RED, GREEN, BLUE matrices into a 2D matrices of red, green, blue color density
dimensions = img.shape

R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]

print(dimensions, R.shape)
# print(R, G, B)
```

```python
# Find the singular value decomposition of matrices R, G, B
from scipy import linalg

U0, s0, V0 = linalg.svd(R)
U1, s1, V1 = linalg.svd(G)
U2, s2, V2 = linalg.svd(B)

print(f'{U0.shape}, {s0.shape}, {V0.shape}')
print(s0, s1, s2)
```

```python
# rank 1 approximation
rank1_appro_R = U0[:, 0].reshape(len(U0[:, 0]), 1) * s0[0] * V0[0, :]
rank1_appro_G = U1[:, 0].reshape(len(U1[:, 0]), 1) * s1[0] * V1[0, :]
rank1_appro_B = U2[:, 0].reshape(len(U2[:, 0]), 1) * s2[0] * V2[0, :]

rank1_appro = np.zeros(dimensions, int)

rank1_appro[:, : ,0] = rank1_appro_R
rank1_appro[:, : ,1] = rank1_appro_G
rank1_appro[:, : ,2] = rank1_appro_B

plt.imshow(rank1_appro)
```

```python
# rank 3 approximation
rank1_appro_R = U0[:, 0].reshape(len(U0[:, 0]), 1) * s0[0] * V0[0, :]
rank1_appro_G = U1[:, 0].reshape(len(U1[:, 0]), 1) * s1[0] * V1[0, :]
rank1_appro_B = U2[:, 0].reshape(len(U2[:, 0]), 1) * s2[0] * V2[0, :]

rank2_appro_R = U0[:, 1].reshape(len(U0[:, 0]), 1) * s0[1] * V0[1, :]
rank2_appro_G = U1[:, 1].reshape(len(U1[:, 0]), 1) * s1[1] * V1[1, :]
rank2_appro_B = U2[:, 1].reshape(len(U2[:, 0]), 1) * s2[1] * V2[1, :]

rank3_appro_R = U0[:, 2].reshape(len(U0[:, 0]), 1) * s0[2] * V0[2, :]
rank3_appro_G = U1[:, 2].reshape(len(U1[:, 0]), 1) * s1[2] * V1[2, :]
rank3_appro_B = U2[:, 2].reshape(len(U2[:, 0]), 1) * s2[2] * V2[2, :]


rank3_appro = np.zeros(dimensions, int)
rank3_appro[:, : ,0] = rank1_appro_R + rank2_appro_R + rank3_appro_R
rank3_appro[:, : ,1] = rank1_appro_G + rank2_appro_G + rank3_appro_G
rank3_appro[:, : ,2] = rank1_appro_B + rank2_appro_B + rank3_appro_B


plt.imshow(rank3_appro)

# polling
```
