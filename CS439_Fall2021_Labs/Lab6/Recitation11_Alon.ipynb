# Gradient of a Linear Model
### Given a dataset $\{(x_i,y_i)\}$, find a function $\hat{y}_i=f_\theta(x_i)$ to approxiate the dataset
### Suppose our model looks like:
### Linear model:
$$
\Large
\hat{y}_i=f_\theta(x_i)=\theta_1^Tx_i+\theta_0
$$
### Loss:
$$
\Large
L(\theta;x,y) = \sum_il(\hat{y}_i,y_i)=\sum_il(f_\theta(x_i),y_i)
$$
### Derivative:
**Using the Chain Rule**

$$
\Large
\frac{\partial l}{\partial \theta_0}=
\frac{\partial l}{\partial \hat{y}_i}\frac{\partial \hat{y}_i}{\partial \theta_0}
+
\frac{\partial l}{\partial y_i}\frac{\partial y_i}{\partial \theta_0}
= \frac{\partial l}{\partial \hat{y}_i}\frac{\partial \hat{y}_i}{\partial \theta_0}
= \frac{\partial l}{\partial \hat{y}_i}\\
$$

\Large

= \frac{\partial l}{\partial \hat{y}_i}\frac{\partial \hat{y}_i}{\partial \theta_1}
= \frac{\partial l}{\partial \hat{y}_i}x_i\\

# Linear Model with Logistic Regression
### Given a dataset $\{(x_i,y_i)\}$, find a function $\hat{y}_i=f_\theta(x_i)$ to approxiate the dataset
### Linear model:
$$
\Large
\hat{y}_i=f_\theta(x_i)=\theta_1^Tx_i+\theta_0
$$
$$
$$
It is possible to put $\theta_0$ in $\theta_1$ to get just $\theta$, by appending the constant value 1 to the end of $x_i$.

Thus, we can write:
$$
$$
$$
\Large
\hat{y}_i=f_\theta(x_i)=\theta^Tx_i
$$
## Logistic Regression

### Here, the answer $y_i$ are {0,1}, and our prediction $\hat{y}_i$ are output probablities (from 0 to 1).

### The function that turns the domain (-inf,inf) to (0,1):

### Sigmoid/Logistic function
$$
\Large
\sigma(h)=\frac{1}{1+e^{-h}}
$$
![sigmoid](sigmoid.png)

### Linear function:
$$
\Large
h_i=f_\theta(x_i)=\theta^Tx_i
$$

### Logistic function:
$$
\Large
\hat {y}_i = P({y}_i=1)=\sigma(h_i)=\frac{1}{1+e^{-\theta^Tx_i}}
$$

### Linear Model with Logistic Regression

```python
# feature engineering and logistic regression
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
# take a look at the input
df = pd.read_csv('train.csv').dropna()
df.head()
```

```python
# input: Age, Sex, Pclass, SibSp
# output: survived

# select input
X = df[['Age', 'Sex', 'Pclass', 'SibSp']]
X['Pclass'] = X['Pclass'].astype('category')
y = df['Survived']

# feature engineering for Sex, Pclass

X_one_hot = pd.get_dummies(X)
X_one_hot.head()
```

```python
# randomly permute X
indices = X_one_hot.sample(frac=1).index
# select some as training, some as testing
X_one_hot_train = X_one_hot.loc[indices[:100]]
X_one_hot_test = X_one_hot.loc[indices[100:]]

y_train = y.loc[indices[:100]]
y_test = y.loc[indices[100:]]
X_one_hot_train
```

```python
log_reg = LogisticRegression()
log_reg.fit(X=X_one_hot_train, y= y_train)
```

```python
import numpy as np
# predict
pred = log_reg.predict(X=X_one_hot_test) 
pred = pd.Series(pred)
pred.index = y_test.index
pred.name = "pred"
print(pd.Series(pred))
print(y_test)
```

```python
merged = pd.concat([pred, y_test], axis=1)
# see how many correct
print('success rate: %f' % (1-np.abs(pred - y_test).mean()))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_true=y_test, y_pred=log_reg.predict(X=X_one_hot_test)))
```
